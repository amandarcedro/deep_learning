```python
# =============================================================================
# PROJETO 2 — Tradução Automática Usando Atenção com LSTM
# =============================================================================
import os
os.environ['PYTHONHASHSEED'] = '0'
os.environ['TF_DETERMINISTIC_OPS'] = '1'
os.environ['TF_CUDNN_DETERMINISTIC'] = '1'
# os.environ['CUDA_VISIBLE_DEVICES'] = ''  # descomente para forçar CPU

import random
import numpy as np
import tensorflow as tf

SEED = 42

def set_seeds(seed=SEED):
    random.seed(seed)
    np.random.seed(seed)
    tf.random.set_seed(seed)
    tf.config.experimental.enable_op_determinism()

set_seeds()
import matplotlib.pyplot as plt
import pickle
import io
import requests

import tensorflow as tf
import tensorflow.keras.backend as K
from tensorflow.keras.layers import (
    Bidirectional, Concatenate, Dot, Input, LSTM,
    RepeatVector, Dense, Activation
)
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.utils import to_categorical

# =============================================================================
# 1. FUNÇÕES AUXILIARES
# =============================================================================

def string_to_int(string, length, vocab):
    """
    Converte uma string em lista de inteiros usando o vocabulário fornecido.
    Usa padding ou truncamento para ajustar ao comprimento desejado.
    """
    string = string.lower()
    string = string.replace(',', '')

    if len(string) > length:
        string = string[:length]

    rep = list(map(lambda x: vocab.get(x, vocab.get('<unk>', 0)), string))

    if len(string) < length:
        rep += [vocab.get('<pad>', 0)] * (length - len(string))

    return rep


def int_to_string(ints, inv_vocab):
    """Converte lista de inteiros de volta para caracteres."""
    return [inv_vocab.get(i, '?') for i in ints]


def softmax_axis1(x):
    """Softmax ao longo do eixo 1 (sobre os passos de tempo de entrada Tx)."""
    return tf.keras.activations.softmax(x, axis=1)


# =============================================================================
# 2. CARREGAMENTO DOS DADOS
# =============================================================================
print("Carregando vocabulários e dataset diretamente do GitHub...")

# URLs no formato 'raw' para download direto dos arquivos binários
urls = {
    'human_vocab': 'https://raw.githubusercontent.com/Amandacedro/Deep_Learning_P2/main/arquivos/human_vocab.pkl',
    'machine_vocab': 'https://raw.githubusercontent.com/Amandacedro/Deep_Learning_P2/main/arquivos/machine_vocab.pkl',
    'inv_machine_vocab': 'https://raw.githubusercontent.com/Amandacedro/Deep_Learning_P2/main/arquivos/inv_machine_vocab.pkl',
    'dataset': 'https://raw.githubusercontent.com/Amandacedro/Deep_Learning_P2/main/arquivos/dataset.pkl'
}

def load_pickle_from_url(url):
    """Baixa o arquivo binário da URL e carrega via pickle."""
    response = requests.get(url)
    response.raise_for_status() # Garante que o download ocorreu com sucesso
    return pickle.load(io.BytesIO(response.content))

# Carregando cada um dos arquivos
human_vocab = load_pickle_from_url(urls['human_vocab'])
machine_vocab = load_pickle_from_url(urls['machine_vocab'])
inv_machine_vocab = load_pickle_from_url(urls['inv_machine_vocab'])
dataset = load_pickle_from_url(urls['dataset'])

print(f"Dataset carregado: {len(dataset)} exemplos.")
print(f"Tamanho do vocabulário humano: {len(human_vocab)}")
print(f"Tamanho do vocabulário de máquina: {len(machine_vocab)}")

# --- Pré-processamento ---
Tx = 30  # comprimento máximo da entrada
Ty = 10  # comprimento fixo da saída (YYYY-MM-DD = 10 caracteres)

X_raw, Y_raw = zip(*dataset)
m = len(dataset)

X = np.array([string_to_int(x, Tx, human_vocab) for x in X_raw])
Y = np.array([string_to_int(y, Ty, machine_vocab) for y in Y_raw])

Xoh = np.array(list(map(lambda x: to_categorical(x, num_classes=len(human_vocab)), X)))
Yoh = np.array(list(map(lambda x: to_categorical(x, num_classes=len(machine_vocab)), Y)))

print(f"\nXoh.shape: {Xoh.shape}")
print(f"\nYoh.shape: {Yoh.shape}")

print(f"\nXoh.shape: {Xoh.shape}")
print(f"Yoh.shape: {Yoh.shape}")

# =============================================================================
# 3. ARQUITETURA DO MODELO DE ATENÇÃO (seguindo estrutura do professor)
# =============================================================================

def build_model(Tx, Ty, n_a, n_s, human_vocab_size, machine_vocab_size):
    """
    Constrói o modelo de atenção com Bi-LSTM + mecanismo de atenção + LSTM pós-atenção.

    Parâmetros:
    -----------
    Tx                 : comprimento da sequência de entrada
    Ty                 : comprimento da sequência de saída
    n_a                : número de unidades da Bi-LSTM (pré-atenção)
    n_s                : número de unidades da LSTM pós-atenção
    human_vocab_size   : tamanho do vocabulário de entrada
    machine_vocab_size : tamanho do vocabulário de saída

    Retorna:
    --------
    model : instância do modelo Keras
    """

    # --- Camadas compartilhadas (seguindo exatamente o padrão do professor) ---
    repeator       = RepeatVector(Tx)
    concatenator   = Concatenate(axis=-1)
    densor1        = Dense(10, activation="tanh")
    densor2        = Dense(1, activation="relu")
    activator      = Activation(softmax_axis1, name='attention_weights')
    dotor          = Dot(axes=1)

    post_activation_LSTM_cell = LSTM(n_s, return_state=True)
    output_layer = Dense(machine_vocab_size, activation=softmax_axis1)

    def one_step_attention(a, s_prev):
        """
        Um passo do mecanismo de atenção.

        Calcula o vetor de contexto como produto ponderado dos estados
        ocultos da Bi-LSTM pelos pesos de atenção (alphas).
        """
        s_prev   = repeator(s_prev)          # (m, Tx, n_s)
        concat   = concatenator([a, s_prev]) # (m, Tx, n_s + 2*n_a)
        e        = densor1(concat)           # energias intermediárias
        energies = densor2(e)               # energias finais (m, Tx, 1)
        alphas   = activator(energies)       # pesos de atenção normalizados
        context  = dotor([alphas, a])        # vetor de contexto (m, 1, 2*n_a)
        return context

    # --- Entradas ---
    X_in = Input(shape=(Tx, human_vocab_size), name='X')
    s0   = Input(shape=(n_s,), name='s0')
    c0   = Input(shape=(n_s,), name='c0')
    s, c = s0, c0

    outputs = []

    # Passo 1: Bi-LSTM pré-atenção — lê a sequência inteira em ambas as direções
    a = Bidirectional(LSTM(n_a, return_sequences=True))(X_in)

    # Passo 2: loop sobre Ty passos decodificador
    for t in range(Ty):
        # 2a. Calcula vetor de contexto pelo mecanismo de atenção
        context = one_step_attention(a, s)

        # 2b. LSTM pós-atenção: atualiza estado oculto s e célula c
        s, _, c = post_activation_LSTM_cell(context, initial_state=[s, c])

        # 2c. Camada densa com softmax para prever o caractere de saída
        out = output_layer(s)
        outputs.append(out)

    # Passo 3: monta o modelo
    model = Model(inputs=[X_in, s0, c0], outputs=outputs)
    return model


# =============================================================================
# 4. CENÁRIO A — MODELO PRÉ-TREINADO DO PROFESSOR
# =============================================================================

print("\n" + "="*60)
print("CENÁRIO A: Carregando modelo pré-treinado...")
print("="*60)

n_a_base = 32
n_s_base = 64

K.clear_session()
model_pretrained = build_model(Tx, Ty, n_a_base, n_s_base,
                               len(human_vocab), len(machine_vocab))

url_model = "https://raw.githubusercontent.com/Amandacedro/Deep_Learning_P2/main/arquivos/model.h5"
local_weight_path = 'model_professor.h5'

try:
    print("Baixando pesos do modelo pré-treinado do GitHub...")
    response_model = requests.get(url_model)
    response_model.raise_for_status() # Garante que o download ocorreu com sucesso

    # Salva o arquivo em disco temporariamente para o Keras ler nativamente
    with open(local_weight_path, 'wb') as f:
        f.write(response_model.content)

    # CARREGA OS PESOS DE FATO NO MODELO
    model_pretrained.load_weights(local_weight_path)
    print("Pesos do modelo pré-treinado carregados com sucesso!")

except Exception as e:
    print(f"Erro ao baixar ou carregar os pesos: {e}")

model_pretrained.summary()

# =============================================================================
# 5. CENÁRIO B — TREINAMENTO DO ZERO: 1 ÉPOCA vs 50 ÉPOCAS
# =============================================================================

s0_train = np.zeros((m, n_s_base))
c0_train = np.zeros((m, n_s_base))
outputs_train = list(Yoh.swapaxes(0, 1))  # lista de Ty arrays (m, machine_vocab_size)

# --- 1 época ---
print("\n" + "="*60)
print("CENÁRIO B: Treinando do zero por 1 época...")
print("="*60)

K.clear_session()
model_1ep = build_model(Tx, Ty, n_a_base, n_s_base,
                        len(human_vocab), len(machine_vocab))
model_1ep.compile(
    optimizer=Adam(learning_rate=0.005, beta_1=0.9, beta_2=0.999, weight_decay=0.01),
    loss='categorical_crossentropy',
    metrics=['accuracy'] * Ty
)
history_1ep = model_1ep.fit(
    [Xoh, s0_train, c0_train], outputs_train,
    epochs=1, batch_size=100
)

# --- 50 épocas ---
print("\n" + "="*60)
print("CENÁRIO C: Treinando do zero por 50 épocas...")
print("="*60)

K.clear_session()
model_50ep = build_model(Tx, Ty, n_a_base, n_s_base,
                         len(human_vocab), len(machine_vocab))
model_50ep.compile(
    optimizer=Adam(learning_rate=0.005, beta_1=0.9, beta_2=0.999, weight_decay=0.01),
    loss='categorical_crossentropy',
    metrics=['accuracy'] * Ty
)
history_50ep = model_50ep.fit(
    [Xoh, s0_train, c0_train], outputs_train,
    epochs=50, batch_size=100
)

# =============================================================================
# 6. FUNÇÃO DE INFERÊNCIA
# =============================================================================

def traduzir(model, data, n_s, verbose=True):
    """
    Realiza a inferência de uma data de formato humano para YYYY-MM-DD.

    Parâmetros:
    -----------
    model   : modelo Keras treinado
    data    : string da data em formato humano (ex: '3 May 1979')
    n_s     : número de unidades do estado oculto da LSTM pós-atenção
    verbose : se True, imprime o resultado

    Retorna:
    --------
    saida : string com a data no formato padronizado
    """
    source = string_to_int(data, Tx, human_vocab)
    source = np.array(list(map(
        lambda x: to_categorical(x, num_classes=len(human_vocab)), source
    ))).swapaxes(0, 1)
    source = np.swapaxes(source, 0, 1)
    source = np.expand_dims(source, axis=0)

    s0_inf = np.zeros((1, n_s))
    c0_inf = np.zeros((1, n_s))

    prediction = model.predict([source, s0_inf, c0_inf], verbose=0)
    prediction = np.argmax(prediction, axis=-1)
    saida = ''.join([inv_machine_vocab[int(i)] for i in prediction])

    if verbose:
        print(f"  Entrada : {data}")
        print(f"  Saída   : {saida}\n")

    return saida


# =============================================================================
# 7. COMPARAÇÃO DOS TRÊS CENÁRIOS
# =============================================================================

EXEMPLOS = [
    '3 May 1979', '5 April 09', '21th of August 2016',
    'Tue 10 Jul 2007', 'Saturday May 9 2018',
    'March 3 2001', '1 March 2001'
]

print("\n" + "="*60)
print("COMPARAÇÃO: Pré-treinado vs 1 época vs 50 épocas")
print("="*60)

for ex in EXEMPLOS:
    print(f"\nEntrada: '{ex}'")
    sa = traduzir(model_pretrained, ex, n_s_base, verbose=False)
    sb = traduzir(model_1ep,        ex, n_s_base, verbose=False)
    sc = traduzir(model_50ep,       ex, n_s_base, verbose=False)
    print(f"  [Pré-treinado] : {sa}")
    print(f"  [1 época]      : {sb}")
    print(f"  [50 épocas]    : {sc}")

# =============================================================================
# 8. ITEM 3 — COMPARAÇÃO DE DIFERENTES VALORES DE n_a E n_s
#    Um único gráfico com acurácia final por configuração
# =============================================================================

print("\n" + "="*60)
print("ITEM 3: Variando n_a (pré-atenção) e n_s (pós-atenção)")
print("="*60)

# Configurações a testar: (n_a, n_s)
configuracoes = [
    (16, 32),
    (32, 32),
    (32, 64),   # configuração base do professor
    (64, 64),
    (64, 128),
]

N_EPOCAS_COMPARACAO = 20  # épocas suficientes para observar diferença sem custo excessivo

resultados = {}  # chave: label, valor: dict com históricos

for (na, ns) in configuracoes:
    label = f"n_a={na}, n_s={ns}"
    print(f"\nTreinando configuração: {label}  ({N_EPOCAS_COMPARACAO} épocas)...")

    s0_cfg   = np.zeros((m, ns))
    c0_cfg   = np.zeros((m, ns))
    out_cfg  = list(Yoh.swapaxes(0, 1))

    K.clear_session()
    modelo_cfg = build_model(Tx, Ty, na, ns, len(human_vocab), len(machine_vocab))
    modelo_cfg.compile(
        optimizer=Adam(learning_rate=0.005, beta_1=0.9, beta_2=0.999, weight_decay=0.01),
        loss='categorical_crossentropy',
        metrics=['accuracy'] * Ty
    )
    hist = modelo_cfg.fit(
        [Xoh, s0_cfg, c0_cfg], out_cfg,
        epochs=N_EPOCAS_COMPARACAO,
        batch_size=100,
        verbose=0   # silencioso para não poluir o output
    )
    # Acurácia média entre as Ty saídas (Keras cria uma chave por saída, não uma global)
    acc_keys = [k for k in hist.history if k.endswith('accuracy') and not k.startswith('val')]
    acc_media = np.mean([hist.history[k] for k in acc_keys], axis=0)

    resultados[label] = {
        'history':   hist,
        'acc_media': acc_media,
        'n_s':       ns
    }
    print(f"  Acurácia média final (época {N_EPOCAS_COMPARACAO}): "
          f"{acc_media[-1]:.4f}  |  Loss: {hist.history['loss'][-1]:.4f}")

# --- Gráfico único: Loss e Acurácia ao longo das épocas para cada configuração ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

for label, res in resultados.items():
    epocas = range(1, N_EPOCAS_COMPARACAO + 1)
    ax1.plot(epocas, res['history'].history['loss'], label=label, marker='o', markersize=3)
    ax2.plot(epocas, res['acc_media'],               label=label, marker='o', markersize=3)

ax1.set_title('Loss por Época — Diferentes Configurações de n_a e n_s')
ax1.set_xlabel('Época')
ax1.set_ylabel('Loss (Categorical Crossentropy)')
ax1.legend(fontsize=8)
ax1.grid(True, alpha=0.3)

ax2.set_title('Acurácia por Época — Diferentes Configurações de n_a e n_s')
ax2.set_xlabel('Época')
ax2.set_ylabel('Acurácia')
ax2.legend(fontsize=8)
ax2.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('comparacao_na_ns.png', bbox_inches='tight')
plt.show()
print("Gráfico salvo em 'comparacao_na_ns.png'.")

# =============================================================================
# 9. ITEM 4 — MAPA DE ATENÇÃO
# =============================================================================

def plot_attention_map(modelx, input_vocabulary, inv_output_vocabulary,
                       text, n_s=64, titulo=None):
    """
    Plota o mapa de atenção para uma data de entrada usando o modelo fornecido.
    Reconstrói a sub-rede de energias seguindo a estrutura do professor.

    Parâmetros:
    -----------
    modelx                : modelo Keras treinado
    input_vocabulary      : dicionário human_vocab
    inv_output_vocabulary : dicionário inv_machine_vocab
    text                  : string de data de entrada
    n_s                   : unidades da LSTM pós-atenção do modelo
    titulo                : título do gráfico (opcional)
    """
    attention_map = np.zeros((Ty, Tx))

    X_in = modelx.inputs[0]
    s0_p = modelx.inputs[1]
    c0_p = modelx.inputs[2]
    s = s0_p
    c = s0_p

    # Reconstrói camada a camada (mesma ordem do build_model / professor)
    a = modelx.layers[2](X_in)
    outputs_energy = []

    for t in range(Ty):
        s_prev   = modelx.layers[3](s)
        concat   = modelx.layers[4]([a, s_prev])
        e        = modelx.layers[5](concat)
        energies = modelx.layers[6](e)
        alphas   = modelx.layers[7](energies)
        context  = modelx.layers[8]([alphas, a])
        s, _, c  = modelx.layers[10](context, initial_state=[s, c])
        outputs_energy.append(alphas)

    f = Model(inputs=[X_in, s0_p, c0_p], outputs=outputs_energy)

    s0_val = np.zeros((1, n_s))
    c0_val = np.zeros((1, n_s))
    encoded = np.array(string_to_int(text, Tx, input_vocabulary)).reshape((1, Tx))
    encoded = np.array(list(map(
        lambda x: to_categorical(x, num_classes=len(input_vocabulary)), encoded
    )))

    r = f([encoded, s0_val, c0_val])

    for t in range(Ty):
        for t_prime in range(Tx):
            attention_map[t][t_prime] = r[t][0, t_prime, 0]

    # Normaliza por linha
    row_max = attention_map.max(axis=1, keepdims=True)
    attention_map = attention_map / (row_max + 1e-8)

    # Predição do modelo para eixo y
    prediction = modelx.predict([encoded, s0_val, c0_val], verbose=0)
    predicted_indices = [int(np.argmax(prediction[t])) for t in range(Ty)]
    predicted_text = int_to_string(predicted_indices, inv_output_vocabulary)

    # Plot
    plt.clf()
    fig = plt.figure(figsize=(8, 8.5))
    ax = fig.add_subplot(1, 1, 1)

    im = ax.imshow(attention_map, interpolation='nearest', cmap='Blues')

    cbaxes = fig.add_axes([0.2, 0, 0.6, 0.03])
    cbar = fig.colorbar(im, cax=cbaxes, orientation='horizontal')
    cbar.ax.set_xlabel('Valor de Alpha (Probabilidade Softmax)', labelpad=2)

    ax.set_yticks(range(Ty))
    ax.set_yticklabels(predicted_text)
    ax.set_xticks(range(len(text)))
    ax.set_xticklabels(list(text), rotation=45)
    ax.set_xlabel('Sequência de Entrada')
    ax.set_ylabel('Sequência de Saída')
    if titulo:
        ax.set_title(titulo)
    ax.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()
    return attention_map


print("\n" + "="*60)
print("ITEM 4: Gerando Mapas de Atenção (modelo treinado — 50 épocas)")
print("="*60)

exemplos_atencao = [
    "Tuesday 09 Oct 1993",
    "3 May 1979",
    "21th of August 2016",
]

fig, axes = plt.subplots(len(exemplos_atencao), 1, figsize=(8, 2 * len(exemplos_atencao)))

ims = []  # guarda os imshow para o colorbar compartilhado

for idx, data in enumerate(exemplos_atencao):
    print(f"\nMapa de atenção para: '{data}'")

    attention_map = np.zeros((Ty, Tx))

    X_in = model_50ep.inputs[0]
    s0_p = model_50ep.inputs[1]
    c0_p = model_50ep.inputs[2]
    s = s0_p
    c = s0_p

    a = model_50ep.layers[2](X_in)
    outputs_energy = []

    for t in range(Ty):
        s_prev   = model_50ep.layers[3](s)
        concat   = model_50ep.layers[4]([a, s_prev])
        e        = model_50ep.layers[5](concat)
        energies = model_50ep.layers[6](e)
        alphas   = model_50ep.layers[7](energies)
        context  = model_50ep.layers[8]([alphas, a])
        s, _, c  = model_50ep.layers[10](context, initial_state=[s, c])
        outputs_energy.append(alphas)

    f = Model(inputs=[X_in, s0_p, c0_p], outputs=outputs_energy)

    s0_val  = np.zeros((1, n_s_base))
    c0_val  = np.zeros((1, n_s_base))
    encoded = np.array(string_to_int(data, Tx, human_vocab)).reshape((1, Tx))
    encoded = np.array(list(map(
        lambda x: to_categorical(x, num_classes=len(human_vocab)), encoded
    )))

    r = f([encoded, s0_val, c0_val])

    for t in range(Ty):
        for t_prime in range(Tx):
            attention_map[t][t_prime] = r[t][0, t_prime, 0]

    row_max = attention_map.max(axis=1, keepdims=True)
    attention_map = attention_map / (row_max + 1e-8)

    prediction = model_50ep.predict([encoded, s0_val, c0_val], verbose=0)
    predicted_indices = [int(np.argmax(prediction[t])) for t in range(Ty)]
    predicted_text = int_to_string(predicted_indices, inv_machine_vocab)

    ax = axes[idx]
    im = ax.imshow(attention_map, interpolation='nearest', cmap='Blues', vmin=0, vmax=1)
    ims.append(im)
    ax.set_yticks(range(Ty))
    ax.set_yticklabels(predicted_text)
    ax.set_xticks(range(len(data)))
    ax.set_xticklabels(list(data), rotation=45)
    ax.grid(True, alpha=0.3)

# Colorbar único — aumenta o pad para não sobrepor os gráficos
fig.colorbar(ims[0], ax=axes, orientation='vertical', fraction=0.046, pad=0.12,
             label='Valor de Alpha (Probabilidade Softmax)')

fig.supxlabel('Sequência de Entrada', fontsize=11)
axes[len(axes)//2].set_ylabel('Sequência de Saída', fontsize=11, labelpad=20)
plt.tight_layout()
plt.subplots_adjust(hspace=0.3, right=0.82)  # right abre espaço para o colorbar
plt.savefig('mapas_atencao.png', bbox_inches='tight')
plt.show()
print("Gráfico salvo em 'mapas_atencao.png'.")
```
